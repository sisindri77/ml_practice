#### 4:

Using only pure python and numpy, implement a simple linear regression algorithm. Use it to derive $w_0$ and $w_1$ for the values below.

```python
import numpy as np
x = np.array([1,2,3])
y = np.array([2.8,5.1,6.7])
```

```python
import matplotlib.pyplot as plt
plt.scatter(
    x, y,
    color='royalblue',
    s=20,
    edgecolors='black'
)

plt.title('Scatter Plot', fontsize=16, fontweight='bold')
plt.xlabel('X Values', fontsize=12)
plt.ylabel('Y Values', fontsize=12)

plt.show()
```

```python
x_mean = np.mean(x)
y_mean = np.mean(y)
x_mean
```

```python
num = np.sum((x - x_mean) * (y - y_mean))
den = np.sum((x - x_mean) ** 2)
w1=num/den
w0=np.mean(y)-(w1*np.mean(x))
print("w0 Intercept = ",w0)
print("w1 slope = ",w1)
```

```python
y_pred = w0 + w1 * x
plt.scatter(x, y, color="blue",s=30,edgecolors="black")
plt.plot(x, y_pred, color="red")
plt.xlabel('X_axis')
plt.ylabel('Y_axis')
plt.text(1.8, 5.5,"y=w0+w1*x")
plt.title("Linear Regression")

```

```python
#Linear regression using sklear
from sklearn.linear_model import LinearRegression
```

```python
#Define model
model=LinearRegression()
#Train the model{fit(x,y)}
fit_result=model.fit(x.reshape(-1,1),y)
#predict(x) — predicts output values.
#intercept_ — returns w0.
#coef_ — returnsw1
w0=fit_result.intercept_
w1=fit_result.coef_
print("intercept w0",w0)
print("coef w1",w1)
```
